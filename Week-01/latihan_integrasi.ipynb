pengenalan macam-macam Library (Numpy, pandas, cv2, skimage, PIL dan
matplotlib)

Langkah (Step) 1:

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

Langkah (Step) 2:

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://placekitten.com/800/571"]

# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  try:
    image = io.imread(url)                            # read image
    image = cv.resize(image, (0, 0), fx=0.5, fy=0.5)  # resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)    # convert color to RGB
    final_frame = cv.hconcat((image, image_2))        # concatenate image
    cv2_imshow(final_frame)                           # show image
    print('\n')

  except Exception as e:
    print("Gagal membaca gambar:", url)
    print("Error:", e)
```

Langkah (Step) 3

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi  x lebar = ",tinggi, " x ", lebar)
cv2_imshow(image_2)
```

Langkah (Step) 4

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2), y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

    Semakin banyak jumlah pixel, semakin tinggi resolusi gambar sehingga gambar dapat menampilkan detail yang lebih banyak. Sebaliknya, semakin sedikit jumlah pixel, semakin rendah resolusi sehingga detail gambar lebih sedikit dan gambar dapat terlihat lebih pecah ketika diperbesar.

```python
# ========================================
# TUGAS 2: Garis Vertikal di Tengah
# ========================================

img_vertikal = image.copy()
h, w, c = img_vertikal.shape

for y in range(h):
    img_vertikal[y, w//2] = [255, 255, 255]

comparison_v = cv.hconcat([image, img_vertikal])
plt.figure(figsize=(12, 6))
plt.imshow(cv.cvtColor(comparison_v, cv.COLOR_BGR2RGB))
plt.title("Kiri: Gambar Asli | Kanan: Garis Vertikal di Tengah", fontsize=14)
plt.axis("off")
plt.show()

# ========================================
# TUGAS 2: Garis Diagonal Menyilang
# ========================================

img_diagonal = image.copy()
h, w, c = img_diagonal.shape

for y in range(h):
    x = int(y * (w - 1) / (h - 1))
    img_diagonal[y, x] = [255, 255, 255]

for y in range(h):
    x = int((h - 1 - y) * (w - 1) / (h - 1))
    img_diagonal[y, x] = [255, 255, 255]

comparison_d = cv.hconcat([image, img_diagonal])
plt.figure(figsize=(12, 6))
plt.imshow(cv.cvtColor(comparison_d, cv.COLOR_BGR2RGB))
plt.title("Kiri: Gambar Asli | Kanan: Garis Diagonal Menyilang", fontsize=14)
plt.axis("off")
plt.show()
```

```python
# ========================================
# TUGAS 3: Garis Horizontal dengan Panjang Tertentu
# ========================================

print("\n" + "="*50)
print("TUGAS 3: Garis Horizontal dengan Panjang Tertentu")
print("="*50)

img_horizontal_panjang = image.copy()
h, w, c = img_horizontal_panjang.shape

panjang = int(w * 0.6)
start_x = int((w - panjang) / 2)
y_tengah = int(h / 2)

for x in range(start_x, start_x + panjang):
    img_horizontal_panjang[y_tengah, x] = [255, 255, 255]

comparison_hp = cv.hconcat([image, img_horizontal_panjang])
plt.figure(figsize=(12, 6))
plt.imshow(cv.cvtColor(comparison_hp, cv.COLOR_BGR2RGB))
plt.title(f"Kiri: Gambar Asli | Kanan: Garis Horizontal (Panjang {panjang} pixel)", fontsize=14)
plt.axis("off")
plt.show()
```

```python
# ========================================
# TUGAS 4: Membuat Kotak dengan Pixel Putih
# ========================================

print("\n" + "="*50)
print("TUGAS 4: Membuat Kotak dengan Pixel Putih")
print("="*50)

img_kotak = image.copy()
h, w, c = img_kotak.shape

ukuran = 150
x_start = int((w - ukuran) / 2)
y_start = int((h - ukuran) / 2)

for y in range(y_start, y_start + ukuran):
    for x in range(x_start, x_start + ukuran):
        img_kotak[y, x] = [255, 255, 255]

comparison_k = cv.hconcat([image, img_kotak])
plt.figure(figsize=(12, 6))
plt.imshow(cv.cvtColor(comparison_k, cv.COLOR_BGR2RGB))
plt.title(f"Kiri: Gambar Asli | Kanan: Kotak Putih ({ukuran}x{ukuran} pixel)", fontsize=14)
plt.axis("off")
plt.show()
```
